#Estatisticas

```python
# Variaveis são as caracteristicas que queremos observar ou medir nos elementos de uma população ou  (caracteristicas)
#Dados: São os valores concretos observados/ medidos para cada variável, em cada elemento.
#Variavel qualitativa : Variavel de adjetivo (qualidade). Não dá pra somar
#Nominal: Categoria sem ordem. EX: cor dos olhos
#Ordinal: Categorias com ordem, que dá pra ordenar. Ex gra de instrução.

#Variavel quantitativa: Trata-se de algum valor numérico.
#discreta: numeros contáveis, geralmente são números inteiros.
#contínua: valores numéricos em uma escala contínua dentro de um intervalo.

times = ["Corinthians", "São Paulo", "Santos"]
# Qualitativa - nominal

numero_filhos = [0,1,1,2,3]
# Quantitativa - discreta

satisfacao = ["péssimo", "ruim", "ótimo","bom"]
# Quaitativa - ordinal

alturas = [1.65, 1.72, 2.02, 1.68]
#quantitativa - contínua
```

#Média

```python
# Soma de valores divididos pela mesma quantidade de números
alturas = [1.65, 1.72, 2.02, 1.68]

media = (1.65+1.72+2.02+1.68) / 4
print(media)

#acumulador
soma = 0
for i in alturas:
  soma = soma + i

medias = soma/len(alturas)
print(medias)

#sum (Função de soma)

soma1 = sum(alturas)
media2 = soma1/len(alturas)
print(media2)
```

#Mediana

```python
#Informação que está no meio de uma lista ordenada

numero_filhos = [0,1,1,2,2,3]

#ordenar

numero_filhos.sort() #sort ordena os dados
tamanho = len(numero_filhos)
meio = 0
mediana = 0
if tamanho % 2 == 1:
  meio = (tamanho / 2 -0.5)-1
  mediana = numero_filhos[int(meio)]

else:
  mediana = (numero_filhos[int(tamanho / 2)] + numero_filhos[int(tamanho / 2 -1)])/2

print(mediana)
```

#MODA

```python
#Busca a informação que mais se repete

satisfacao = ["péssimo", "ruim", "ruim","ótimo","bom" , "bom" ,"bom"]

dicionario_satisfacao = {}

for satis in satisfacao: # Roda variável satis na lista
  if satis in dicionario_satisfacao.keys(): #Compara a variavel satis com o dicionário
    dicionario_satisfacao[satis] += 1 # Acrescenta 1 de valor se a chave for existente
  else:
    dicionario_satisfacao[satis] = 1 # Cria a chave e adiciona 1 no valor

print(dicionario_satisfacao)

moda = ""
maior = 0
for c,v in dicionario_satisfacao.items():
  if v > maior: # Verifica o valor se o valor for maior no dicionário.
    maior = v
    moda = c

print(f"A moda é {moda} e repetiu {maior} vezes")

```
